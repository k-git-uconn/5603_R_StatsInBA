```r
# create a vector for weight
# the c() just means 'combine' so we are combining many numbers to make a vector
weight <- c(60, 72, 57, 90, 95, 72) # green means numeric
# show values for weight
weight
```

```r
# add val1, val2, val3 and call it val4
val4 <- val1 + val2 + val3
val4
```

```r
# create a vector for height
height <- c(1.75, 1.80, 1.65, 1.90, 1.74, 1.91)
# show values for height
height
```

# Combining two variables with a formula
We will use a formula to make a new variable `bmi` which we can play with later on (rounding, plotting, etc.)

```r
# create a new variable from those two
# R will obey order of operations
bmi <- weight/(height^2)
bmi
```

# Useful functions
sum(), round()

```r
# round bmi to 3 decimal points
# on your own, try rounding to 5 or 0 and note the difference
round(bmi, 3)
```

```r
# you can store it for later
roundedBMI <- round(bmi, 3)
roundedBMI
```

```r
# sum all values in the vector "weight"
sum(weight)
```

The `length()` function is useful for counting the number of entries in a vector.

```r
# this is a way we can calculate the mean of the vector
# sum all the values then divide by the count (or length) of the vector
sum(weight)/length(weight)
```

```r
# here's a better way (using a function!)
mean(weight)
```

```r
# and we can round it too
round(mean(weight), 2) # Try 3 or 5 here

# see how we put a function in a function?
```

```r
# we can assign it to another variable
xbar <- sum(weight)/length(weight)
```

```r
# this is the actual value minus the mean value
weight - mean(weight)
```

```r
# actual minus the mean, then squared (deviation from the mean, squared)
(weight - mean(weight))^2
```

# Data types
R has 5 data types but we are concentrating on only 3 for now: numeric, character, and logical.

Data Types matter because they behave differently - you can only do math on numeric/float/integer data - you can't do math on a string or a logical!

```r
# check the class of weight
class(weight)
```

# Strings
Strings are characters (and character is a 'data type') - strings have quotation marks around them. If you put quotation marks around a number, it becomes a string. You can't do math on a string!

```r
# here's a vector of strings
HDL = c("Huey","Dewey","Louie")
class(HDL)
```

# Logicals
A logical is a data type. Typically it is data that's stored as 0 or 1; Yes or No; or True or False (you may come across all of these in the class).

```r
# here's a vector of LOGICALS
# note that there aren't any quotation marks - it's not a string!
TF <- c(T,T,F,T)
class(TF)
```

```r
# here's another way
# again - note that there aern't any quotation marks!
c(TRUE, TRUE, FALSE, TRUE)

# note that this will make an error because R is case sensitive
#c(True, True, False, True)
```

```r
# here's a 'logical test' applied to a vector
# it's evaluating where the condition is true in the vector
bmi > 25
```

```r
# as a newbie, you should ALWAYS be verifying your answer -
# we see that indeed it's true!
bmi
```

# Subsetting a vector


```r
# a vector of numbers
intake.pre <- c(5260,5470,5640,6180,6390,
                6515,6805,7515,7515,8230,8770)
# another vector of numbers
intake.post <- c(3910,4220,3885,5160,5645,
                 4680,5265,5975,4680,5265,5975)
```

```r
# return the 5th row entry of the vector
intake.pre[5]
```

```r
# return the 3rd, 5th and 7th row entries
intake.pre[c(3,5,7)]
```

```r
# we can use other variables to subset... advanced!
v <- c(3,5,7)
intake.pre[v]
```

```r
# we can use a range (:) to subset
intake.pre[c(1:5)]
intake.pre[1:5] # this works too...
```

```r
# - is convient! but be careful...
# this mean return everything BUT the 3rd, 5th and 7th rows
intake.pre[-c(3,5,7)]
```

```r
# this is how we subset values > 7000... it selects where 'TRUE' by default!
print('Intake Post')
intake.post
print('Intake Pre')
intake.pre
# subset
print('Subset')
intake.post[intake.pre > 7000]
```

```r
# we can combine functions just like excel, but using & ("ampersand")
intake.post[intake.pre > 7000 & intake.pre <= 8000]
```

```r
# to make this clear, look - this returns a logical!
intake.pre > 7000 & intake.pre <= 8000
```
