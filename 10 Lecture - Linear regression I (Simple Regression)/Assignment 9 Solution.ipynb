# Problem 1 (35 Points)

One concern in Major League Baseball is that the time to play games has become too long. There have been recent efforts to enforce rules that will speed up the game. In the data set ```BaseballGames.csv``` we have data reporting a set of games that were played on a given day, where the columns, in order, report the game index, the teams playing the game, the league (either AL or NL), the margin of victory, the number of pitchers, the attendance at the game, and the amount of time that the game took, in minutes. We will conduct a statistical test to determine if the average length of games is less than 200 minutes at the 95% confidence level.

a. State the null hypothesis.

Comment:

Null Hypothesis:

**H0: mu >= 200**


The average MLB game length is 200 minutes or longer.

b. State the alternative hypothesis.

Comment:

H1 : mu < 200

The average MLB game length is less than 200 minutes.
- Significance level: (alpha=0.05)
- Test type: One-tailed, lower-tail test
- Confidence level: 95%

c. Run a parametric statistical
test. What is the p-value for the statistical test? Show all steps.

````
filepath_BaseballGames = "https://raw.githubusercontent.com/k-git-uconn/5603_R_StatsInBA/main/10%20Lecture%20-%20Linear%20regression%20I%20(Simple%20Regression)/BaseballGame.csv"
baseball_df <- read.csv(filepath_BaseballGames)

# Check the dataset
head(baseball_df)
names(baseball_df)
str(baseball_df)
````

````
# Extract game time (last column)
game_time = baseball_df$Time

# Sample statistics
sample_size = length(game_time)
sample_mean = mean(game_time)
sample_sigma = sd(game_time)

population_mean = 200
alpha = 0.05

SE = sample_sigma / sqrt(sample_size)

cat("Sample size =", sample_size, "\n")
cat("Sample mean =", sample_mean, "\n")
cat("Sample standard deviation =", sample_sigma, "\n")

t_sample = (sample_mean - population_mean) / SE
cat("t statistic =", t_sample, "\n")

p_value = pt(t_sample, df=sample_size-1)
cat("p-value =", p_value, "\n")

t_value = qt(alpha, df=sample_size-1)
cat("Critical t-value =", t_value, "\n")
````

Comment:

Decision


1. p-value	(0.08415 > 0.05)   | 	Fail to reject H0
2. Critical value	(-1.452893 > -1.76131) |	Fail to reject H0

At the 5% significance level, we fail to reject the null hypothesis because the p-value (0.08415) is greater than 0.05, and the calculated t-statistic (-1.452893) is greater than the critical t-value (-1.76131).

Therefore, there is insufficient statistical evidence to conclude that the average MLB game length is less than 200 minutes.

d. At a confidence level of 95%, would you reject or fail to reject the null hypothesis?

a. Please conduct a two-sample test to check if there is sufficient statistical evidence to conclude that the mean ```basePay``` of female workers is different from that of male workers? State the null hypothesis and the alternative hypothesis.

Hypotheses
- H₀: μFemale = μMale (the mean basePay is equal for female and male workers).
- H₁: μFemale ≠ μMale (the mean basePay is different).


This is a two-tailed, independent two-sample t-test.

b. Please provide the test statistics for question (a) and draw your conclusion.

````
# Separate basePay by gender
basePay_female = glassdoor_df$basePay[glassdoor_df$gender == "Female"]
basePay_male = glassdoor_df$basePay[glassdoor_df$gender == "Male"]

# Sample statistics
n_female = length(basePay_female)
n_male = length(basePay_male)

mean_female = mean(basePay_female)
mean_male = mean(basePay_male)

sd_female = sd(basePay_female)
sd_male = sd(basePay_male)

# Welch two-sample t-test
SE = sqrt(sd_female^2/n_female + sd_male^2/n_male)

t_sample = (mean_female - mean_male)/SE

df = (sd_female^2/n_female + sd_male^2/n_male)^2 /
  ((sd_female^2/n_female)^2/(n_female-1) +
   (sd_male^2/n_male)^2/(n_male-1))

p_value = 2*pt(-abs(t_sample), df=df)

cat("Female sample size =", n_female, "\n")
cat("Male sample size =", n_male, "\n")
cat("Female mean basePay =", mean_female, "\n")
cat("Male mean basePay =", mean_male, "\n")
cat("t statistic =", t_sample, "\n")
cat("Degrees of freedom =", df, "\n")
cat("p-value =", p_value, "\n")
````

- Degrees of freedom = 991.2189
- p-value = 0.00000008719
- Significance level = 0.05
Reject H₀
Assignment-ready conclusion:

Since the p-value (8.718919 × 10⁻⁸) is much smaller than the significance level of 0.05, we reject the null hypothesis.

There is sufficient statistical evidence to conclude that the mean basePay of female workers is significantly different from that of male workers.

In this sample, female workers have a lower average basePay ($89,942.82) than male workers ($98,457.55), a difference of approximately $8,514.73.

Hypotheses

- H₀: μIT = μData Scientist (the mean bonus is equal for both job titles).
- H₁: μIT ≠ μData Scientist (the mean bonus is different).


This is also a two-tailed, independent two-sample t-test.

d. Please provide the test statistics for question (c) and draw your conclusion.

````
# Separate bonus by job title
bonus_IT = glassdoor_df$bonus[glassdoor_df$jobTitle == "IT"]
bonus_DS = glassdoor_df$bonus[glassdoor_df$jobTitle == "Data Scientist"]

# Sample statistics
n_IT = length(bonus_IT)
n_DS = length(bonus_DS)

mean_IT = mean(bonus_IT)
mean_DS = mean(bonus_DS)

sd_IT = sd(bonus_IT)
sd_DS = sd(bonus_DS)

# Welch two-sample t-test
SE = sqrt(sd_IT^2/n_IT + sd_DS^2/n_DS)

t_sample = (mean_IT - mean_DS)/SE

df = (sd_IT^2/n_IT + sd_DS^2/n_DS)^2 /
  ((sd_IT^2/n_IT)^2/(n_IT-1) +
   (sd_DS^2/n_DS)^2/(n_DS-1))

p_value = 2*pt(-abs(t_sample), df=df)

cat("IT sample size =", n_IT, "\n")
cat("Data Scientist sample size =", n_DS, "\n")
cat("IT mean bonus =", mean_IT, "\n")
cat("Data Scientist mean bonus =", mean_DS, "\n")
cat("t statistic =", t_sample, "\n")
cat("Degrees of freedom =", df, "\n")
cat("p-value =", p_value, "\n")
````

# Problem 3 (30 Points)

A random sample of $20$ items from the first population showed a mean of $100$ and a standard deviation of $15$. A sample of $16$ items for the second population showed a mean of $94$ and a standard deviation of $8$. Use the $0.05$ significant level. Assume the sample populations do not have equal standard deviations.



````
# Sample information
sample_size1 = 20
sample_size2 = 16

sample_mean1 = 100
sample_mean2 = 94

sample_sigma1 = 15
sample_sigma2 = 8

alpha = 0.05

#Assume the sample populations do not have equal standard deviations.
````

a. Write the formula for the number of degrees of freedom. Explain what everything in the formula is. Then compute the degrees of freedom for this problem.

For unequal population variances, use the Welch-Satterthwaite formula:

Unknown and unequal sigma


  df = (sd_1^2/n_1 + sd_2^2/n_2)^2/((sd_1^2/n_1)^2/(n_1-1) + (sd_2^2/n_2)^2/(n_2-1))


  df = trunc(df)

Explanation of variables:
- \(sd_1\) = standard deviation of sample 1 (15).
- \(sd_2\) = standard deviation of sample 2 (8).
- \(n_1\) = sample size of population 1 (20).
- \(n_2\) = sample size of population 2 (16).
- \(df\) = approximate degrees of freedom.

````
# Calculate standard error
SE = sqrt(sample_sigma1^2/sample_size1 +
          sample_sigma2^2/sample_size2)

# Calculate degrees of freedom
df = (sample_sigma1^2/sample_size1 +
      sample_sigma2^2/sample_size2)^2 /
     ((sample_sigma1^2/sample_size1)^2/(sample_size1-1) +
      (sample_sigma2^2/sample_size2)^2/(sample_size2-1))

# Calculate t statistic
t_sample = (sample_mean1-sample_mean2)/SE

# Critical t-value
t_value = qt(1-alpha/2, df=df)

# Two-tailed p-value
p_value = 2*pt(-abs(t_sample), df=df)

cat("SE =", SE, "\n")
cat("Degrees of freedom =", df, "\n")
cat("t statistic =", t_sample, "\n")
cat("Critical t-value =", t_value, "\n")
cat("p-value =", p_value, "\n")
````
