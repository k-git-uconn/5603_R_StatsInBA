# Plotting and Tables

## Scatter plot
Let’s use mtcars

```r
# Attach dataset for later use
# The mtcars dataset is included in the base R distribution, specifically within the datasets package. This package is automatically loaded when you start an R session, so you can use mtcars without needing to load any additional packages.
attach(mtcars)
summary(mtcars)
```

Let’s build a scatter plot of weight and mpg

```r
# This plots a scatter plot of wt on mpg
plot(wt,mpg)
```

```r
# This adds the labels for the x and y axis
plot(wt,mpg, xlab = "Weight (1000s of pounds)", ylab = "Miles per gallon")
```

```r
# Add title to the plot using the main argument
# Or try to change the point shape using pch
plot(wt,mpg,main = "Miles per gallon versus weight", xlab = "Weight (1000s of pounds)", ylab = "Miles per gallon", pch = 3)

# Common pch Values:
# pch = 1: An open circle (default).
# pch = 2: An open triangle.
# pch = 3: A plus sign.
# pch = 4: A cross.
# pch = 5: An open diamond.
# pch = 6: An open inverted triangle.
# pch = 7: An open square with a plus sign inside.
# pch = 8: A star.
# pch = 16: A solid circle.
# pch = 17: A solid triangle.
# pch = 18: A solid diamond.
# pch = 19: A larger solid circle.
# pch = 20: A smaller solid circle.
# pch = 21 to pch = 25: Symbols with both filled and outlined components, where the filling color can be controlled by bg.
```

```r
# Change the argument of pch from 3 to 20
# The pch argument allows you to specify which symbol to use for the points.
plot(wt,mpg,main = "Miles per gallon versus weight", xlab = "Weight (1000s of pounds)", ylab = "Miles per gallon", pch = 20)
```

# Boxplots

Lets do boxplots.

```r
# basic function
boxplot(mpg)
```

```r
# Add color to the box plot
boxplot(mpg, col = "yellow")
```

```r
#http://www.stat.columbia.edu/~tzheng/files/Rcolor.pdf
# Can do it horizontally
boxplot(mpg, col = "blue", horizontal =  TRUE)
```

```r
# Annotate the plot
boxplot(mpg, col = "blue", horizontal = TRUE, axes = FALSE)
text(x=fivenum(mpg), labels =fivenum(mpg), y=1.03)
text(x=fivenum(mpg), labels =c("Min","Q1","Q2","Q3","MAX"), y=0.9)

# Arguments x and y specify the coordinates where the text should be placed on a plot.
# For example, x = 3, y = 5: Places the text at the point (3, 5) on the plot.
```

```r
# Annotate the plot by shortening the distance between two rows of texts through y values
boxplot(mpg, col = "blue", horizontal =  TRUE, axes = FALSE)
text(x=fivenum(mpg), labels =fivenum(mpg), y=1.3)
text(x=fivenum(mpg), labels =c("Min","Q1","Q2","Q3","MAX"), y=1.25)
```

# Histogram

Let’s do histogram

```r
# Histogram plot for the variable age
hist(wt)
```

```r
# Change the number of breaks
hist(wt,breaks = 5)
```

```r
# Change the color of the histogram to green
hist(wt,breaks = 5, col = "green")
```

```r
# Change the color of each bar of the histogram
ABC <- c("blue","green","red","orange","yellow","purple")
hist(wt,breaks = 5, col = ABC)
```

```r
# Specify color using number
hist(wt,breaks = 5, col = c(1:6))
```

Specify the breaks

```r
# Specify the breaks
myBreaks4 <- seq(min(wt),max(wt),length=4)
myBreaks3 <- seq(min(wt),max(wt),length=3)
print(myBreaks4)
print(myBreaks3)
```

```r
# Histograms with myBreak3 as breaks
hist(wt,breaks = myBreaks3)
```

```r
# Histograms with myBreak4 as breaks
hist(wt,breaks = myBreaks4)
```

```r
#Use x lim is to specify the limits of the x-axis in the plot. This allows you to control the range of values displayed on the x-axis, effectively zooming in or out on a particular region of the data.
#help(hist)
hist(wt,breaks = myBreaks4, xlim = c(1,6))
```

```r
# Distribution with frequency
# Frequency: The sum of the heights of all bars equals the total number of data points.
hist(wt,breaks = myBreaks4, xlim = c(1,6), freq = FALSE)
hist(wt,breaks = myBreaks4, xlim = c(1,6),freq = TRUE)
```

```r
hist(wt,breaks = myBreaks4, xlim = c(1,6),freq = TRUE,col = "blue")
```
