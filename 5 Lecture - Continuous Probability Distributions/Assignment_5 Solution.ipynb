# Problem 1 (30 Points):

Suppose that the time it takes for a worker to complete a task is modeled as a uniform random variable.

a. Suppose that the minimum value for the uniform distribution is assumed to be 19 and that the maximum value is assumed to be 26. What is the probability that the time it takes a worker to complete a task is greater than or equal to 23 minutes?


```
#Given
#Uniform Random Variable
#minimum value = 19
#maximum value = 26
#ask = prob of work complete is >=23 minutes

min = 19
max = 26
#P(X>=23)
#1-P(X<23)

prob_lessthan23 = 1 - punif(23, min, max)
prob_lessthan23
```

# Problem 2 (30 Points)

The temperature of coffee sold at the Coffee Bean Cafe follows the normal probability distribution, with a mean of 150 degrees. The standard deviation of this distribution is 5 degrees.

a. What is the probability that the coffee temperature is between 150 degrees and 159 degrees?


```
#Given
mu = 150
sd = 5

#ask P(150 < X < 159)

prob_ask = pnorm(159, 150, 5) - pnorm(150, 150, 5)
prob_ask
```

b. What is the probability that the coffee temperature is more than 154 degrees?

```
#ask P(X>154)

#1-P(X<=154)

prob_ask = 1 - pnorm(154, mean = 150, sd = 5)

prob_ask
```

c. If I know that the coffee temperature is between 150 degrees and 159 degrees, what is the proability that the the coffee temperature is more than 154 degrees?

```
#we know
#P(150 < X < 159)

#ask is
#P(X > 154) = P(154 < X < 159)

#P(X>154 | 150 < X < 159) = P(X>154) / P(150 < X < 159) = P(154 < X < 159) / P(150 < X < 159)

upper_fraction = pnorm(159, mean = 150, sd = 5) - pnorm(154, mean = 150, sd = 5)
upper_fraction

lower_fraction = pnorm(159, mean = 150, sd = 5) - pnorm(150, mean = 150, sd = 5)
lower_fraction

prob_ask = upper_fraction / lower_fraction

prob_ask
```
