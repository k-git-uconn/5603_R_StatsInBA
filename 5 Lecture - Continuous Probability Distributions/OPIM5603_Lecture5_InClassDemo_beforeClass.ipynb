5. What is the probability that X is greater than 57 conditional on X lies between 55 and 65?

We are interested in $$P(X > 57|55 < X < 65) = \frac{P(X > 57\,and\, 55 < X < 65)}{P(55 < X < 65)} = \frac{P(57 < X < 65)}{P(55 < X < 65)})$$

```r
#P(57<X<65)
probability_lessthan65 <- pnorm(65, mean = 60, sd = 5)
probability_lessthan57 <- pnorm(57, mean = 60, sd = 5)

#P(55<X<65)
probability_lessthan55 <- pnorm(55, mean = 60, sd = 5)

ask_prob = (probability_lessthan65 - probability_lessthan57) / (probability_lessthan65 - probability_lessthan55)
ask_prob
```

# Slide 15: Standard Normal Distribution


## Converting to z-values
Any normal probability distribution can be converted into a standard normal
probability distribution by subtracting the mean from each observation and dividing
this difference by the standard deviation. **The results are called z values or z scores.** <br>
<center>$Z=$$(\frac{X-\mu}{\sigma})$</center>

So, a z value is the distance from the mean, measured in units of the standard
deviation. The Z distribution has mean = 0 and standard deviation = 1.

The formula for the standardized normal probability density function is SIMPLIFIED from what we see above:

<center>$f(Z) =$ $\frac{1}{\sqrt2\pi}e^{\frac{-1}{2}Z^{2}}$</center>

Where:
* $e$ = the mathematical constant approximated by 2.71828
* $\pi$ = the mathematical constant approximated by 3.14159
* $Z$ = any value of the standardized normal distribution

ANY normal distribution can be converted to the standard normal distribution.




## Example 1: Standard Normal Distribution
Suppose we wish to compute the probability that boxes of Sugar Yummies weigh between 283 and 285.4 grams. We know that the box weight of Sugar Yummies follows the normal distribution with a mean of 283 grams and a standard deviation of 1.6 grams.

1. We want to know the probability or area under the curve between the mean, 283 grams, and 285.4 grams.

We will show how you can solve this problem two ways: a) with pnorm() and b) by converting to the standard normal distribution.

Of course, we can use pnorm() to solve this quickly. We'll do this first.

```r
# Use pnorm to solve it directly
#P(283<X<285.4)

probab_283 = pnorm(283, 283, 1.6)
probab_283
probab_285_4 = pnorm(285.4, 283, 1.6)
probab_285_4

final_probab = probab_285_4 - probab_283
final_probab
```

We can now solve the problem using the standard normal distribution. It requires us to us the z-formula (shown below again, for clarity).

<center>$Z=$$(\frac{X-\mu}{\sigma})$</center>

We can express this
problem using probability notation, similar to the style used in the previous chapter:
$P(283 < weight < 285.4)$. To find the probability, it is necessary to convert both 283
grams and 285.4 grams to z values using formula.

The **z value corresponding to
283 is 0**, found by $\frac{(283-283)}{1.6}$.

The **z value corresponding to 285.4 is 1.50** found
by $\frac{(285.4-283)}{1.6}$.

```r
# The answer should match the results in the previous code cell

#P(283<weight<285.4)

#convert to standard normal distribution
z_283 = (283-283) / 1.6
z_283
z_285_4 = (285.4-283) / 1.6
z_285_4

pnorm(z_285_4) - pnorm(z_283)


```

# Slide 18: Uniform Distribution
The continuous uniform distribution (also called a rectangular distribution) is the probability distribution of random number selection from the continuous interval between **a** and **b**. Its density function is defined by the
following.<br>

\begin{equation}
f(x)=\begin{cases}\frac{1}{b-a} & if\,a \le x \le b\\ 0 & otherwise \end{cases}
\end{equation}

where:
$f(x)$ = value of the density function at any x value
$a$ = minimum value of x
$b$ = maximum value of x


The mean of a uniform distribution is:<br>
<center>$\mu$ = $\frac{a+b}{2}$</center>

The standard deviation is:<br>
<center>$\sigma$ = $\sqrt(\frac{(b-a)^2}{12})$</center>




Suppose $X \sim \text{Uniform}[2,6]$

1. What is $P(3 \leq X \leq 5)?$

P(3 < X < 5) = F(5)- F(3)

```r
# Use two punifs
# F(5)- F(3)

p_3 = punif(3, 2, 6)
p_3
p_5 = punif(5, 2, 6)
p_5

p_53 = p_5 - p_3
p_53
```

2. What is the probability that X lie between -9 and 5?

P(-9 < X < 5) = F(5) - F(-9)


```r
# Use two punifs
p_9 = punif(-9, 2, 6)
p_9
p_5 = punif(5, 2, 6)
p_5

p_95 = p_9 - p_5
p_95
```
