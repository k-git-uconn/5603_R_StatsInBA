# Problem 1 (35 Points)

Let $X$ be a random variable that is normally distributed with mean 8 and standard deviation 4. Let $Y$ be a random variable that is normally distributed with mean 10 and standard deviation 2. $X$ and $Y$ are independent of each other. Let $Z$ be a random variable that is $5X - 3Y$.

Please set the number of replication = 1000 for any problems that required simulation and use the command set.seed(202210).



a. What is the expected value of $X$?

Comment: E[X] = mean = 8

b. What is the expected value of $Y$?

Comment: E[Y] = mean = 10

c. What is the expected value of $Z$?

Comment:

Given: Z = 5X - 3Y

E[Z] = E[5X-3Y] => 5E[X]-3E[Y]

=> 5(8)-3(10) = 40-30
=> 10

So, E[Z] = 10

d. What is the variance of $Y + Z$?

Comment:

Variance of \(Y+Z\)


Since \(Z=5X-3Y\), So
Y+Z => Y+5X-3Y => 5X-2Y

Because \(X\) and \(Y\) are independent, their covariance is zero.

Also Var(aX) = a^2 Var(X)


Therefore,

\begin{aligned}
\operatorname{Var}(Y+Z) => 5^2\operatorname{Var}(X)+(-2)^2\operatorname{Var}(Y)
&=> 25\operatorname{Var}(X)+4\operatorname{Var}(Y)\\
&=> 25(4^2)+4(2^2)\\
&=> 400+16 => 416
\end{aligned}



Variance of $Y + Z$ = 416


e. Use simulation to estimate the correlation of $Y$ and $Z$. What is your estimate of the correlation?

```
set.seed(202210)

replication <- 1000

X <- rnorm(replication, mean = 8, sd = 4)
Y <- rnorm(replication, mean = 10, sd = 2)

Z <- 5 * X - 3 * Y

cor(Y, Z)

# Why is it negative? Since \(Z=5X-3Y\), increasing \(Y\) decreases \(Z\), holding \(X\) constant.
```

f. Simulate $Y+Z$. What is the estimated variance of $Y+Z$ based on the simulation?

```
Y_plus_Z <- Y + Z

var(Y_plus_Z)
```

# Problem 3 (30 points)

The owner of the West End Kwick Fill Gas Station wishes to determine the proportion of customers who use a credit card or debit card to pay at the pump. He surveys 100 customers and finds that 80 paid at the pump.


Comment:

Given:
- Sample size: n => 100
- Customers who paid at the pump: x => 80
- Sample proportion: p_cap => 80 / 100 => 0.80


a. What is the point estimate the value of the population proportion?

```
x = 80
n = 100

p_hat = x / n
p_hat
```

b. What is the 95 percent confidence interval for the population proportion?

```
x = 80
n = 100

p_hat = x / n

confidence <- 0.95

# Critical z-value
z_score <- qnorm(1 - (1 - confidence)/2)
z_score

SE = sqrt(p_hat * (1 - p_hat) / n)
ME = z_score * SE

lower = p_hat - ME
upper = p_hat + ME

c(lower, upper)
```

c. Interpret your findings.

Comment: We are 95% confident that the true proportion of all West End Kwick Fill customers who pay at the pump using a credit or debit card is between 72.16% and 87.84%.

The sample estimate is 80%, suggesting that most customers pay at the pump.
