a. A sample of size $n = 200$ produced the sample mean of $\bar{X} = 63$. Assuming the population standard deviation $\sigma= 5$, compute a 95% confidence interval for the population mean $\mu$.

```
# Given value for population
sigma = 5
confidence = 0.95

# Given values for sample
n = 200
xbar = 63


# Critical z-value
z = qnorm(1 - (1 - confidence)/2)
# OR we can round of two digit
z_95 = 1.96

# Standard error
SE = sigma / sqrt(n)

# Margin of error
ME = z_95 * SE

# Confidence interval
lower = xbar - ME
upper = xbar + ME

c(lower, upper)
```

When population standard deviation (sigma) is known, use qnorm() and the Z-confidence interval.

# Problem 2 (50 points)

Suppose that you are interested in estimating the proportion of computers from a manufacturer that have faulty motherboards. Please answer each subpart independent of the others.

a. Suppose you take a random sample of size 250, and 40 have faulty motherboards. Build a 90% confidence interval for the proportion of all computer that have this defect. You can assume that $n · \pi \geq 5$ and that $n · (1 - \pi) \geq 5$.

```
# Given information for sample
# size
n = 250
# faulty motherboards
x = 40
# confidence level 90
confidence = 0.90

p_hat = x / n

# Critical z-value
z = qnorm(1 - (1 - confidence)/2)
z
z_90 = 1.65
z_90

# Standard Error
SE = sqrt(p_hat * (1 - p_hat) / n)
# Marginal Error
ME = z_90 * SE

lower = p_hat - ME
upper = p_hat + ME

c(lower, upper)
```

Comment: We are 90% confident that the true proportion of all computers with faulty motherboards is between 12.19% and 19.82%.

b. Suppose you take a random sample of size 170, and 25 have faulty motherboards. Build a 95% confidence interval for the proportion of all computer that have this defect. You can assume that $n · \pi \geq 5$ and that $n · (1 - \pi) \geq 5$.

```
# Given information for sample
# size
n = 170
# faulty motherboards
x = 25
# confidence level 90
confidence = 0.95

p_hat = x / n

# Critical z-value
z = qnorm(1 - (1 - confidence)/2)
z
z_95 = 1.96
z_95

# Standard Error
SE = sqrt(p_hat * (1 - p_hat) / n)
# Marginal Error
ME = z_95 * SE

lower = p_hat - ME
upper = p_hat + ME

c(lower, upper)
```

c. Suppose that the true population parameter $\pi$ is 0.25. Replicate 1000 samples of size 200, and look at the distribution of the sample proportions. Does it look like the sampling distribution of the sample proportion is normal? Use a histogram, and discuss.

```
set.seed(123)

# population parameter pi
pi = 0.25

# size
n = 200

# replicate samples count
num_samples = 1000

# size = 1, In our problem, each computer has two possible outcomes:
# 1 = Faulty motherboard (25% probability).
# 0 = No faulty motherboard (75% probability).

sample_proportions = replicate(num_samples, {
  sample_data = rbinom(n, size = 1, prob = pi)
  mean(sample_data)
})

hist(sample_proportions,
     main = "Sampling Distribution of Sample Proportions",
     xlab = "Sample Proportion",
     col = "lightblue",
     border = "black")

mean(sample_proportions)
sd(sample_proportions)
```

Bell-shaped distribution: The histogram is roughly symmetric, with most sample proportions concentrated around 0.25.



Center: The distribution is centered approximately at 0.25, which matches the true population proportion (pi=0.25).



Spread: Most sample proportions fall approximately between 0.20 and 0.30.



Normality: The distribution looks approximately normal, with fewer observations toward both tails.
