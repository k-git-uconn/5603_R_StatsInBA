# Choosing an Appropriate Sample Size

There are three factors that determine the sample size when we wish to estimate the population parameter:<br>
1. The margin of error the researcher will tolerate.
2. The level of confidence desired, for example, 95%.
3. The variation or dispersion of the population being studied.

# Sample Size to Estimate a Population Mean
To estimate a population mean, we can express the interaction among these three
factors and the sample size in the following formula. Notice that this formula is the margin of error used to calculate the endpoints of confidence intervals to estimate a population mean!<br>
<center>$E=z\frac{\sigma}{\sqrt n}$</center><br>


Solving the above equation yields, the formula for Sample Size to Estimate the Population mean which is: <br>
<center>$n = \left(\frac{Z\sigma}{E}\right)^2$ </center><br>

where:<br>
$n$ is the size of the sample.<br>
$z$ is the standard normal value corresponding to the desired level of confidence.<br>
$\sigma$ is the population standard deviation.<br>
$E$ is the maximum allowable error.

##Slide 20

A student in public administration wants to determine the mean amount members of
city councils in large cities earn per month as remuneration for being a council member. The error in estimating the mean is to be less than \$100 with a 95 percent level of confidence. The student found a report by the Department of Labor that reported a standard deviation of \$1,000. What is the required sample size?

```
E <- 100
sigma <- 1000
z <- qnorm(1-(1-0.95)/2)
z

#Or
z_975 <- qnorm(0.975)
z_975
```

```
n <- (z * sigma / E)^2

ceiling(n)
```

# Sample Size to Estimate a Population Proportion

To determine the sample size for a proportion, the same three variables need to be
specified:
1. The margin of error.
2. The desired level of confidence.
3. The variation or dispersion of the population being studied.

For the binomial distribution, the margin of error is:<br>
<center> $E = z\sqrt{\frac{\pi*(1-\pi)}{n}}$</center><br>

Solving this equation for n yields the following equation for the Sample Size for the Population Proportion:<br>

<center> $n=\pi(1-\pi)(\frac{Z}{E})^2$</center><br>

where:<br>
n is the size of the sample<br>
z is the standard normal value corresponding to the desired level of confidence<br>
$\pi$ is the population proportion<br>
E is the maximum allowable error.

## Slide 42 A Two-tailed test example

Problem: Jamestown Steel Company manufactures
and assembles desks and other office equipment at several plants in western New York state. The weekly production of the Model
$A325$ desk at the Fredonia Plant follows a
normal probability distribution with a mean of
$200$ and a standard deviation of $16$. Recently,
because of market expansion, new production methods have been introduced and new
employees hired. The vice president of manufacturing would like to investigate whether there has been a change in the weekly production of the Model A325 desk. The mean number of desks produced last year ($50$ weeks, because the plant was shut down $2$ weeks for vacation) is $203.5$. Is the mean number of desks produced at the Fredonia Plant different from $200$ at the $0.05$ significance level?

```
# Given the population properties are....

z_value = qnorm(1-0.05/2) # for population
z_value

# Given the sample is...

z_value_sample = (203.5-200)/(16/sqrt(50))
z_value_sample

# Conclusion: Because 1.55 is smaller than 1.95 and does not lies in Rejection area.
# Hence H0 null hypothesis is accepted.
```

OR

```
alpha <- 0.05

qnorm(alpha/2)
qnorm(1-alpha/2)
```

```
xbar <- 203.5
mu <- 200
sigma <- 16
n <- 50

z <- (xbar - mu)/(sigma/sqrt(n))
z
```

## Slide 46 Example
Suppose instead of wanting to know if there had been a change in the mean number of desks assembled, the vice president wants to know whether there has been an *increase* in the number of units assembled. Can we conclude, because of the improved production methods, that the mean number of desks assembled in the last $50$ weeks was more than $200$?

```
# Given the population properties are..

z_value = qnorm(0.95) # for population ***** Note this is the major difference between one tail and two tail tests
z_value

# Given the sample is...

xbar <- 203.5
mu <- 200
sigma <- 16
n <- 50

z <- (xbar - mu) / (sigma / sqrt(n))
z
# 1.5468
```

# Hypothesis Testing: Example

# Slide 70 Example

Consider a random sample of 10 salaries (in $ thousands): 62, 71, 75, 47, 85, 85, 82, 78, 52, 53. Assume that the salary is normally distributed. Conduct a statistical test to determine if the mean salary is greater than 60 at 5 percent significance level. What is the p-value of this test?

```
# Construct the dataset
tmp <- c(62, 71, 75, 47, 85, 85, 82, 78, 52, 53)


sample_mean = mean(tmp)
sample_mean
sample_sigma = sd(tmp)
sample_sigma
```

```
# for the population
# alpha 0.05 then it is a 95% confidence interval
# H0: pi <= 60 (upper one tail), right tail
# H1: pi > 60
# t_value = qt(?)

# for the sample
t_sample = (sample_mean - 60) / (sample_sigma/sqrt(10))
t_sample
```

```
t_value = qt(1-0.05, df=10-1)
t_value
```

# t_sample > t_value, so we are going to reject.

```
# df (degree of freedom = 10-1)
pt(t_sample, df = 9, lower.tail = FALSE)
```

# Tests Concerning Proportions

## Slide 58 Example<br>
Suppose prior elections in a certain state indicated it is necessary for a candidate for governor to receive at least 80 percent of the vote in the northern section of the state to be elected. The incumbent governor is interested in assessing his chances of returning to office and plans to conduct a survey of 2,000 registered voters in the northern section of the state. 1550 voters say they will vote for him.
Using the hypothesis-testing procedure, assess the governor’s chances of
reelection at 0.05 significance level.

```
# H0: pi >= 80%
# H1: pi < 80%
# alpha = 0.05 then confidence interval is 95%
z_value <- qnorm(0.05)
z_value

# Find z_X
p = 1550 / 2000
n = 2000
piee = 0.80

z_X = (p - piee) / sqrt(piee * (1-piee) /n)
z_X
```

z_value is larger than z_X, so we are going to reject H0.

```
# LETS TRY THE OTHER WAY

# H0: pi <= 80%
# H1: pi > 80%
# alpha = 0.05 then confidence interval is 95%
z_value <- qnorm(1-0.05)
z_value

# Find z_X
p = 1550 / 2000
n = 2000
piee = 0.80

z_X = (p - piee) / sqrt(piee * (1-piee) /n)
z_X
```

# Slide 70 Example

Consider a random sample of 10 salaries (in $ thousands): 62, 71, 75, 47, 85, 85, 82, 78, 52, 53. Assume that the salary is normally distributed. Conduct a statistical test to determine if the mean salary is greater than 60 at 5 percent significance level. What is the p-value of this test?

```
# Construct the dataset
tmp <- c(62, 71, 75, 47, 85, 85, 82, 78, 52, 53)
```

```
# for the population
# alpha 0.05 then it is a 95% confidence interval
# H0: pi <= 60 (upper one tail)
# H1: pi > 60
t_value = qt(1-0.05, df = 10-1)
t_value

# for the sample
sample_mean = mean(tmp)
sample_mean

sample_sigma = sd(tmp)
sample_sigma

n = 10

t_x = (sample_mean-60) / (sample_sigma / sqrt(n))
t_x

# Right tail: 1.96 > 1.83, reject H0
```

```
# Alternative testing using p-value

1-pt(t_x, 9)

# At significance level of 5%, p-value being 0.04 (4%) is smaller than 5%, therefore, we reject H0
```
