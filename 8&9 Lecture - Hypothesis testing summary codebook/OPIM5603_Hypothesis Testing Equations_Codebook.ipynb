# Confidence Interval


```
## Confidence Intervals to Estimate a Population Mean

  # (1) Known sigma
  SE = sigma / sqrt(n)
  Critical_value = qnorm(1-alpha/2)
  upper_bound = mean + SE * Critical_value
  lower_bound = mean - SE * Critical_value

  # (2) Unknown sigma
  SE = sample_sd / sqrt(n)
  Critical_value = qt(1-alpha/2,df=n-1)
  upper_bound = mean + SE * Critical_value
  lower_bound = mean - SE * Critical_value

## Confidence Intervals to Estimate a Population Proportion

  SE = sqrt(mean * (1 - mean) / n)
  Critical_value = qnorm(1-alpha/2)
  upper_bound = mean + SE * Critical_value
  lower_bound = mean - SE * Critical_value
```
