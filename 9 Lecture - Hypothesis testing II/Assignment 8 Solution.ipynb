# Problem 1 (60 Points)
A hardware store is considering restructuring their shelving and the proposed plan will only have space for 162 propane tanks to be on display each day. Assume that the hardware store does not have worker capacity to restock shelves during a day, but will always restock up to the capacity for
the start of the business day. In a random sample of 32 days, the mean number of tanks sold is 150.8 and the standard deviation is 50.3. The company drafting the plans claims that the hardware
store has enough shelving to satisfy average daily demand for propane tanks. We will conduct a statistical test to evaluate this claim. Please use H0: mu >= 162 as your null hypothesis

a. Formulate the null hypothesis (in words) to evaluate this claim.

**Comment:** The average daily demand for propane tanks at the hardware store is at least 162 tanks per day, meaning the proposed shelving capacity of 162 tanks is sufficient to meet average daily demand.

So, null hypothesis is:
**H0: mu >= 162**

b. Formulate the alternative hypothesis (in words) to evaluate this claim.

**Comment:**

Alternative hypothesis is:
**H1: mu < 162**

This is a left-tailed test because we are testing whether average daily demand is below 162 tanks.

c. What is $\mu_0$ in the statistical test?

**Comment:**

**$\mu_0$ = 162**

The hypothesized population mean is 162 propane tanks per day, which is the maximum daily shelving capacity.

d. What is the population parameter that the statistical test is concerned with?

**Comment:** The population parameter is $\mu$ (mu), which represents the true average number of propane tanks sold per day at the hardware store across all business days.

\\[ \boxed{\bar{x}=150.8\text{ tanks per day}} \\]

f. Is using the parametric tests discussed in class applicable? Explain why or why not.

**Comment:** Yes, using a parametric test is applicable, specifically a one-sample t-test.

Explanation:
- We are testing a population mean ($\mu$).
- The sample size is 32, which is reasonably large.
- The population standard deviation is unknown, but the sample standard deviation (50.3) is available.

Therefore, **a one-sample, left-tailed** is appropriate to determine whether average daily demand is less than 162 propane tanks.

g. What values of $\bar{x}$ would lead you to reject the null hypothesis at a confidence level of 90%? Calculate this based on the t-test and z-test.

```
# H0: mean >= 162
# H1: mean < 162
# Lower one-tailed test

# Given the population properties are...
population_mean = 162

# Given the sample properties...
sample_mean = 150.8
sample_sigma = 50.3
sample_size = 32

alpha = 0.10 # (1 - 0.90)

# Standard Error
SE = sample_sigma / sqrt(sample_size)

# t-test: Critical value
t_value = qt(alpha, df = sample_size - 1)
#t_value

# Critical sample mean for t-test
t_sample = population_mean + (t_value * SE)
#t_sample

cat("t_value =", t_value, "\n")
cat("t_sample =", t_sample, "\n")

# z-test: Critical value
z_value = qnorm(alpha)
#z_value

# Critical sample mean for z-test
z_sample = population_mean + (z_value * SE)
#z_sample

cat("z_value =", z_value, "\n")
cat("z_sample =", z_sample, "\n")

# Compare the sample mean with critical sample means
sample_mean
```

Conclusion: Both tests produce similar results. The t-test is more appropriate because the population standard deviation is unknown. The actual sample mean is 150.8, which is above both critical values. Therefore, we fail to reject the null hypothesis at the 90% confidence level.

c. Declare if this a two-tail or one-tail test? If it's a one-tail, is this an upper tail or lower tail test? What is your decision rule?

**Comment:** This is a one-tailed test, specifically a right-tailed test.

Because the question asks whether the mean number of customers is more than 100 per day

```
# H0: mu <= 100
# H1: mu > 100

# Given:
# Significance level: alpha = 0.05
# Sample size: n = 15
# Degrees of freedom: df = 15-1 = 14

alpha = 0.05
sample_size = 15

# Upper-tailed t-test
t_value = qt(1-alpha, df=sample_size-1)

cat("Critical t-value =", t_value, "\n")
```

Decision rule:
- If t_sample > 1.761, reject H0.
- If t_sample <= 1.761, fail to reject H0.

d. Solve the problem numerically. (using R code). Calculate the t-value corresponding to your answer.

```

# Self-checkout customers at Walmart
tmp = c(120, 108, 120, 114, 118, 91, 118, 92,
        104, 104, 112, 97, 118, 108, 117)

# Hypotheses
# H0: mu <= 100
# H1: mu > 100
# Upper one-tailed test

# Given population properties
population_mean = 100

# Given sample properties
sample_size = length(tmp)
sample_mean = mean(tmp)
sample_sigma = sd(tmp)

cat("sample_size =", sample_size, "\n")
cat("sample_mean =", sample_mean, "\n")
cat("sample_sigma =", sample_sigma, "\n")

# Standard Error
SE = sample_sigma / sqrt(sample_size)
cat("SE =", SE, "\n")

# Calculate observed t-statistic
t_sample = (sample_mean - population_mean) / SE
cat("t_sample =", t_sample, "\n")

# Critical t-value
alpha = 0.05
t_value = qt(1-alpha, df=sample_size-1)
cat("t_value =", t_value, "\n")

# Calculate p-value (upper-tailed)
p_value = pt(t_sample, df=sample_size-1, lower.tail=FALSE)
cat("p_value =", p_value, "\n")
```

# Problem 3 (20 Points)

In recent years, the interest rate on home mortgages has declined to less than 6.0 percent. However, according to a study by the Federal Reserve Board, the rate charged on credit card debt is more than 14 percent. Listed below is the interest rate charged on a sample of 10 credit cards.

`14.6 16.7 17.4 17.0 17.8 15.4 13.1 15.8 14.3 14.5`

Is it reasonable to conclude the mean rate charged is greater than 14 percent? Use the 0.01 significance level.

We need to determine whether the mean interest rate charged on credit card debt is greater than 14% using a 1% significance level.

Since we have a sample of 10 credit cards and the population standard deviation is unknown, we will use a one-sample t-test.

a. Write the null and alternate hypothesis.

**Comment:**

Null hypothesis, **H0: mean <= 14**


Alternative hypothesis, **H1: mean > 14**

b. Write the formula you are going to use to calculate the test statistic.

**Comment:**

x_bar => Sample mean

mu = 14 => Hypothesized population mean

(sd) => Sample standard deviation

(n=10) => Sample size

Substituting the known values: t = (x_bar-14) / (sd / sqrt(10))

c. Declare if this a two-tail or one-tail test? If it's a one-tail, is this an upper tail or lower tail test? What is your decision rule?

**Comment:** This is a one-tail (right-tail) test because the Alternative hypothesis, **H1: mean > 14**

Decision rule:

At significance level (alpha=0.01), with (df => n-1 => 9):

- Reject H₀ if t_sample > t_critical.
- Fail to reject H₀ if t_sample <= t_critical.

d. Solve the problem numerically. (using R code). Calculate the p-value corresponding to your answer. State your decision about rejecting the null hypothesis.


```

# Credit card interest rates
tmp = c(14.6, 16.7, 17.4, 17.0, 17.8, 15.4, 13.1, 15.8, 14.3, 14.5)

# H0: mean <= 14
# H1: mean > 14
# Upper one-tailed test

# Given population properties
population_mean = 14

# Given sample properties
sample_size = length(tmp)
sample_mean = mean(tmp)
sample_sigma = sd(tmp)

alpha = 0.01

# Standard Error
SE = sample_sigma / sqrt(sample_size)

# Calculate observed t-statistic
t_sample = (sample_mean - population_mean) / SE

# Critical t-value
t_value = qt(1-alpha, df=sample_size-1)

# Calculate p-value (upper-tailed)
p_value = pt(t_sample, df=sample_size-1,
             lower.tail=FALSE)

# Display results
cat("sample_size =", sample_size, "\n")
cat("sample_mean =", sample_mean, "\n")
cat("sample_sigma =", sample_sigma, "\n")
cat("SE =", SE, "\n")
cat("t_sample =", t_sample, "\n")
cat("t_value =", t_value, "\n")
cat("p_value =", p_value, "\n")
```

**Comment:** At the 1% significance level, we reject the null hypothesis because the calculated t-statistic (3.4008) is greater than the critical t-value (2.8214). Additionally, the p-value (0.00393) is smaller than the significance level of 0.01.

Therefore, there is sufficient statistical evidence to conclude that the mean interest rate charged on credit card debt is greater than 14%.
