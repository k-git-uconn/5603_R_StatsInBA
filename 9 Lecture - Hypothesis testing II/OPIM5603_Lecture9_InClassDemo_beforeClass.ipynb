# Two-Sample Tests of Hypothesis: Independent Samples



The variance of the difference in sample means ($\bar{X_1} - \bar{X_2}$) is equal to the sum of the variance for the plumbers and the variance for the electricians.

We can write the formula for the **variance of the distribution of differences in means** is calculated by the formula:<br>
<center> $\sigma^2_{\bar{X_1}-\bar{X_2}} = \frac{\sigma^2_1}{n_1} + \frac{\sigma^2_2}{n_2}$ <br></center>

But we don't use the variance of the distribution of means directly (although it is good to know!) Instead, we can put this equation in a more usable form by taking the square root, so
that we have the standard deviation of the distribution or “standard error” of the differences.

Another way we can say this is 'what is the standard deviation of the distribution of differences?'

 We standardize the distribution of the differences. The result is the following equation: <br>

Two sample test of means with known $\sigma$ is given by:
<center> $Z$ = $\frac{{\bar{X_1}-\bar{X_2}}}{\sqrt{\frac{\sigma^2_1}{n_1} + \frac{\sigma^2_2}{n_2}}}$ <br></center>

# Comparing Population Means with Unknown Population Standard Deviations

## Equal Population Standard Deviations

The following formula is used to pool the sample standard deviations:<br>
<center> $s_p^2$ = $\frac{(n_1-1)s_1^2 + (n_2-1)s_2^2}{n_1+n_2-2}$</center>

where:<br>
$s_1^2$ is the variance (standard deviation squared) of the first sample.<br>
$s_2^2$ is the variance of the second sample.<br>

The value of $t$ is computed from the following equation.

Two-Sample tests of Means - Unknown Standard Deviation:<br>
<center> $t$ = $\frac{{\bar{X_1}-\bar{X_2}}}{\sqrt{s_p^2(\frac{1}{n_1} + \frac{1}{n_2}})}$ <br></center>

where:<br>
$\bar{X_1}$ is the mean of the first sample<br>
$\bar{X_2}$ is the mean of the second sample<br>
$n_1$ is the number of observations in the first sample<br>
$n_2$ is the number of observations in the second sample<br>
$s_p^2$ is the pooled estimate of the population variance

The number of degrees of freedom in the test is the total number of items sampled minus the total number of samples. Because there are two samples, there are $n_1 + n_2 -2$ degrees of freedom.

## Unequal Population Standard Deviations

The formula for the test statistic $t$ for no difference in means, unequal variances:<br>
<center> $t$ = $\frac{{\bar{X_1}-\bar{X_2}}}{\sqrt{\frac{s^2_1}{n_1} + \frac{s^2_2}{n_2}}}$ <br></center>

The degrees of freedom for unequal variance test is found by:<br>
<center> $df$ = $\frac{[(\frac{s_1^2}{n_1}) + (\frac{s_2^2}{n_2})]^2}{\frac{(\frac{s_1^2}{n_1})^2}{n_1 - 1} + \frac{(\frac{s_2^2}{n_2})^2}{n_2 - 1}}$</center><br>
where,<br>
$n_1$ and $n_2$ are the respective sample sizes<br>
$s_1$ and $s_2$ are the respective sample standard deviations<br>
If necessary, this fraction is rounded down to an integer value.

### Slide 15 Example
Personnel in a consumer testing laboratory are evaluating the absorbency of
paper towels. They wish to compare a set of store brand towels to a similar group of name brand ones. An independent random sample of 12 name brand towels absorbed the following

```
store_brand = c(8, 8, 3, 1, 9, 7, 5, 5, 12)

name_brand = c(12, 11, 10, 6, 8, 9, 9, 10, 11, 9, 8, 10)

alpha = 0.10

n_s <- length(store_brand)
n_n <- length(name_brand)

x_s <- mean(store_brand)
x_n <- mean(name_brand)

sd_s <- sd(store_brand)
sd_n <- sd(name_brand)

sd_two_samples = sqrt((sd_s^2/n_s) + (sd_n^2/n_n))
sd_two_samples


df <- (sd_s^2/n_s + sd_n^2/n_n)^2 /
      (((sd_s^2/n_s)^2/(n_s-1)) +
       ((sd_n^2/n_n)^2/(n_n-1)))

df = trunc(df)
df

#Now, do the two sample tests
#H0: mu_name = mu_store


t_value = qt(1-alpha/2, df)
t_value

t_x = (x_s - x_n) / sd_two_samples
t_x

# +-2.4 is outside +-1.82, so H0 is rejected
```

# Two-Sample Tests of Hypothesis: Dependent Samples


## Slide 19 Example
Recall that Nickel Savings and Loan wishes to compare the two companies it uses
to appraise the value of residential homes. Nickel Savings selected a sample of $10$ residential properties and scheduled both firms for an appraisal. The results, reported in $\$000$, are:

image.png

At the $0.05$ significance level, can we conclude there is a difference in the mean appraised values of the homes?

Let \(d = Schadek appraisal - Bowyer appraisal)
- \( H0: mu_d = 0 )     (no difference in average appraisals)
- \( H1: mu_d neq= 0 )  (there is a difference in average appraisals)


This is a two-tailed test with (alpha=0.05).

```
schadek <- c(235, 210, 231, 242, 205,
             230, 231, 210, 225, 249)

bowyer <- c(228, 205, 219, 240, 198,
            223, 227, 215, 222, 245)

# Calculate paired differences
d = schadek - bowyer

# Sample size and summary statistics
n = length(d)
sample_mean = mean(d)
sample_sigma = sd(d)

cat("Differences =", d, "\n")
cat("Sample size =", n, "\n")
cat("Mean difference =", sample_mean, "\n")
cat("Standard deviation =", sample_sigma, "\n")

#################################################

population_mean = 0
SE = sample_sigma / sqrt(n)

t_sample = (sample_mean - population_mean) / SE

cat("SE =", SE, "\n")
cat("t statistic =", t_sample, "\n")

#################################################

alpha = 0.05
t_value = qt(1-alpha/2, df=n-1)

cat("Lower critical value =", -t_value, "\n")
cat("Upper critical value =", t_value, "\n")

#################################################


p_value = 2 * pt(-abs(t_sample), df=n-1)

cat("p-value =", p_value, "\n")

#################################################

CI_lower = sample_mean - t_value * SE
CI_upper = sample_mean + t_value * SE

cat("Lower CI =", CI_lower, "\n")
cat("Upper CI =", CI_upper, "\n")
```

```
d <- schadek - bowyer

mean(d)
sd(d)
```

Since:
- \(t=3.3045\) is greater than the critical value \(2.2622\).
- p-value \(=0.00916\) is less than \(\alpha=0.05\).
- The confidence interval does not include zero.


We reject the null hypothesis.

There is sufficient statistical evidence that the two appraisal companies have different mean appraisals. Schadek's average appraisal is approximately $4,600 higher than Bowyer's.

# Comparing Dependent and Independent Samples

# Slide 20 Example

To illustrate, we will use the Nickel Savings and Loan example just completed. Suppose we assume that we have two independent samples of real estate property for appraisal and conduct the following test of hypothesis.

```
schadek <- c(235, 210, 231, 242, 205,
             230, 231, 210, 225, 249)

bowyer <- c(228, 205, 219, 240, 198,
            223, 227, 215, 222, 245)

# Independent two-sample t-test
t.test(schadek, bowyer,
       paired = FALSE,
       alternative = "two.sided",
       var.equal = FALSE,
       conf.level = 0.95)
```

Notice something interesting:

Test	P-value	Decision at 5%


Paired t-test	0.00916	Reject H₀


Independent t-test	0.4833	Fail to reject H₀



The same numerical data produce very different results because the statistical methods use different assumptions.
