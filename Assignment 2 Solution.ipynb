````
n = length(Diameter)
print(paste("Total length of vector Diameter is:", n))
mean.Diameter = sum(Diameter)/n
print(paste("Mean is: ", mean.Diameter))
cal.sd = sqrt(sum((Diameter - mean.Diameter)^2) / (n-1))
print(paste("Standard deviation: ", cal.sd))
````

Hint: In the slides for Lecture 2, I have shown you another way to calculate geometric mean with the ```log()``` and ```exp()``` functions.

````
#Step 1
logx = log(randomValues)

#Step 2
sumx = sum(logx)
meanx = sumx / length(randomValues)

#Step 3
geomean = exp(meanx)
geomean
````
