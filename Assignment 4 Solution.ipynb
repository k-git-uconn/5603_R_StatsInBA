# Problem 2 (35 points)—Binomial Distribution

The owner of a car dealership is looking to model the number of purchases of vehicles. Let $X$ be the number of purchases of vehicles in a single day, assuming that 90 customers arrive on that day. Historical data suggests that a customer will purchase a vehicle with probability 0.15. Answer the following questions with R. In Problem 2, you are not required to proof the answer for each question. Instead, you should use R code to generate a variable that answers each of the following subquestions and then print it. Please show all work and add some necessary explanation for each step.

a. What is the expected value of $X$?

```
n <- 90
p <- 0.15

expected_value <- n * p
print(expected_value)
```

b. What is the variance of $X$?

```
n <- 90
p <- 0.15

variance <- n * p * (1 - p)
print(variance)
```

c. What is the probability that 12 cars are purchased on a single day?

```
n <- 90
p <- 0.15

probability_12 <- dbinom(12, size = n, prob = p)
print(probability_12)
```


d. What is the probability that 16 or fewer cars are purchased on a single day?

```
n <- 90
p <- 0.15

probability_12 <- dbinom(12, size = n, prob = p)
print(probability_12)
```

The probability that exactly 12 cars are purchased on a single day is approximately 11.10%.

e. What is the probability that at least 25 cars are purchased on a single day?

```
n <- 90
p <- 0.15

probability_25_or_more <- 1 - pbinom(24, size = n, prob = p)
print(probability_25_or_more)
```

Or we can use lower.tail = FALSE

```
#P(X=>25)

pbinom(24, size = 90, prob = 0.15, lower.tail = FALSE)
```

f. What is the probability that less than 19 cars are purchased on a single day?

```
#P(X<19)

probability_less_19 <- pbinom(18, size = n, prob = p)
print(probability_less_19)
```

We use pbinom(18, ...) because R's pbinom() calculates the probability of getting 18 or fewer successes, which is exactly the same as getting fewer than 19 cars.

g. What is the probability that more than 22 cars are purchased on a single day?

```
probability_more_22 <- pbinom(22, size = n, prob = p, lower.tail = FALSE)
print(probability_more_22)
```

Why lower.tail = FALSE?

Setting lower.tail = FALSE calculates the complement.

h. What is the probability that the number of cars purchased on a single day is between 15 and 22 (including of 15 and 22)?

```
probability_15_to_22 <- pbinom(22, size = n, prob = p) -
                        pbinom(14, size = n, prob = p)

print(probability_15_to_22)
```

Because the question includes both 15 and 22, we use pbinom(22) and subtract everything up through 14 using pbinom(14).

i. What is the probability that the number of cars purchased on a single day is not strictly between 12 and 18 (excluding 12 and 18)?

```
#P(13<=X<=17)

#P(not strictly between 12 an 18) = 1 - P(13<=X<=17)

probability_not_between <- 1 - (pbinom(17, size = n, prob = p) -
                                pbinom(12, size = n, prob = p))

print(probability_not_between)
```

Because 12 and 18 are excluded from the "strictly between" range, they are included in the answer.

j. What is the probability that 14 or more cars are purchased on a day?

```
probability_14_or_more <- 1 - pbinom(13, size = n, prob = p)
print(probability_14_or_more)
```

We use pbinom(13) because “14 or more” starts at 14, so its complement is 13 or fewer.
