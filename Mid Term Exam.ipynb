## Problem 1a
If events A and B are independent, then A' (Complement) and B are independent. (Here A' (Complement) means the complement of A).

Answer: TRUE

Reason: If A and B are independent, then we know
P(AnB) = P(A and B) = P(A) . P(B)

To prove: then A (Complement) and B are independent.
We know that P(A'nB) = P(A' and B) = P(A').P(B)

Then P(A' and B) = P(A').P(B)

which drive to P(A'nB) = P(A').P(B)

## Problem 1b
Two mutually exclusive and collectively exhaustive events are complements.

Answer: TRUE

Reason: If A and B are mutually exclusive (NO two events can occur simultaneously) and collectively exhaustive (If one of the event must occur) events,

then P(AnB) = 0


and P(AuB) = P(A) + P(B) - P(AnB) = P(A) + P(B)

Replace P(B) = 1-P(A)


P(AuB) = P(A) + 1 - P(A) = 1

So that means, two events are complements to each other.

Denominator: [P(T|I) * P(I)] + [P(T|I') * P(I')] =
 = 0.99 * 0.1 +  0.05 * 0.99
 = 0.099 + 0.0495
 = 0.1485

P(I|T) = 0.099 / 0.1485 = 0.66

Unsuccessful event probability:

P(S1') = 0.3, P(S2') = 0.2, P(S3') = 0.1

Two events:

event A = all three investment successful

event B = atleast one of investment successful

## Solution 6

For independent events,
P(AuB) = P(A) + P(B)

When all successful events
P(A) = P(S1)*P(S2)*P(S3)
=> (0.7)(0.8)(0.9) = 0.504

From: P(S1) = 0.7, P(S2) = 0.8, P(S3) = 0.9

Also unsuccessful events we evaluated earlier using complement rule:

 P(S1') = 0.3, P(S2') = 0.2, P(S3') = 0.1

When atleast one successful events, So we evaluate: Sum of all 3 below kind events:

1. exact one successful event
P(B=1)

=> P(S1)*P(S2')*P(S3') + P(S1')*P(S2)*P(S3') + P(S1')*P(S2')*P(S3)

=> (0.7)(0.2)(0.1) + (0.3)(0.8)(0.1) + (0.3)(0.20(0.9)

=> 0.092

2. exact two successful event
P(B=2)

=> P(S1')*P(S2)*P(S3) + P(S1)*P(S2')*P(S3) + P(S1)*P(S2)*P(S3')

=> (0.3)(0.8)(0.9) + (0.7)(0.2)(0.9) + (0.7)(0.8)(0.1)

=> 0.398

3. exact three successful event
P(B=3)

=> P(S1)*P(S2)*P(S3) = (0.7*0.8*0.9) = 0.504

So, prob of atleast one successful investment =  

P(B) => P(B=1) + P(B=2) + P(B=3) = 0.092 + 0.398 + 0.504 = 0.994

Ask is P(AuB) = P(A) * P(B)

=> 0.504 * 0.994 = 0.50 (approx)
